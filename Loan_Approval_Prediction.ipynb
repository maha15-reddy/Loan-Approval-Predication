# 1.Business Understanding

### problem statement:
* A bank wants to predict whether a loan application should be approved based on the applicant's

presonal and financial information.This helps reduce manual effort and speeds up the loan approval proces

### Target Variable

Loan_status:
* y=Loan Approved
* N=loan Rejected

## Business Objectives:

* Reduce Manual Verification
* Minimize the risk of approval high-risk applicants
* Improve Customer Experience

## Machine Learning Problem

* Type: Supervised learning
* Category : Binary Classification
* Target Variable: Loan_Status

# 2.Data Understanding

These are the main Features/Columns available in the dataset :

1. No_of_dependents – Number of dependents the applicant has
2. Education – Graduate / Not Graduate
3. Self_employed – Whether the applicant is self-employed
4. Income_annum – Annual income of the applicant
5. Loan_amount – Loan amount requested
6. Loan_term – Loan repayment time (years)
7. Cibil_score – Credit score indicating creditworthiness
8. Residential_assets_value – Value of residential assets
9. Commercial_assets_value – Value of commercial assets
10. Luxury_assets_value – Value of luxury assets
11. Bank_asset_value – Bank balance / financial assets
12. Loan_status (Target variable) – Approved (1) or Rejected (0)

# 10.Hyperparameter Tuning

Hyperparameter tuning is the process of finding best parameter values for ML model to improve its performance

* Improve Model Accuracy
* Reduce Overfitting
* Finds best model automatically

### GridserchCV

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [3, 10, 15],
    'min_samples_split': [2, 5, 10],
}
```

```python
# Apply Grid

rf=RandomForestClassifier(random_state=42)

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

grid_search.fit(x_train, y_train)
```

```python
print(grid_search.best_params_)
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_
y_pred=best_model.predict(x_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

### RandomizedSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
param_grid = {
    'n_estimators': [50, 100, 159, 200, 250, 300],
    'max_depth': [5, 10, 15, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf':[1,2,4]
}
```

```python
random_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_distributions=param_grid,
    n_iter=20,
    cv=5,
    scoring='accuracy',
    random_state=42,
    n_jobs=-1
)

random_search.fit(x_train, y_train)
```

```python
# Best Parameters

print(random_search.best_params_)
print(random_search.best_score_)
```

```python
best_model = random_search.best_estimator_
y_pred_best = best_model.predict(x_test)
```

```python
accuracy = accuracy_score(y_test, y_pred_best)
print("Accuracy:", accuracy)
```

### Roc Curve

```python
from sklearn.metrics import roc_curve
from sklearn.metrics import roc_auc_score

y_prob = best_model.predict_proba(x_test)[:,1]
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label='ROC Curve')
plt.plot([0,1],[0,1],'r--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

### Auc curve

```python
auc = roc_auc_score(y_test, y_prob)

print("AUC Score:", auc)
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_prob)
plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, color='blue',label=f'ROC Curve (AUC = {auc:.3f})')

plt.plot([0,1], [0,1], linestyle='--', color='red', label='Random Classifier')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve with AUC")
plt.legend(loc='lower right')

plt.show()

print("AUC Score:", round(auc, 3))
```
